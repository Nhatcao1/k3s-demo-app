# HE SDK — FIDES GPU image

The Docker image already contains patched OpenFHE, FIDESlib, the native Python binding, and `he_looming_sdk`. Run the cells in order; no installation or compilation is required.

```python
import subprocess
from he_sdk import HESession, __version__
import he_sdk_fides

subprocess.run(["nvidia-smi"], check=True)
print("he_sdk:", __version__)
print("he_sdk_fides:", he_sdk_fides.__version__)
```

## Create one GPU session

```python
session = HESession.create(device="gpu")
print("backend:", session.capabilities.backend)
print("operations:", session.capabilities.operations)
```

## Encrypt inputs

```python
left_values = [1.0, 2.0, 3.0, 4.0]
right_values = [10.0, 20.0, 30.0, 40.0]

left_ct = session.encrypt(left_values)
right_ct = session.encrypt(right_values)

print("left plaintext:", left_values)
print("left ciphertext:", left_ct)
print("left decrypted:", session.decrypt(left_ct))
```

## Encrypted arithmetic

```python
add_ct = session.add(left_ct, right_ct)
subtract_ct = session.subtract(left_ct, right_ct)
multiply_ct = session.multiply(left_ct, right_ct)
square_ct = session.square(left_ct)

print("add expected/decrypted:", [11.0, 22.0, 33.0, 44.0], session.decrypt(add_ct))
print("subtract expected/decrypted:", [-9.0, -18.0, -27.0, -36.0], session.decrypt(subtract_ct))
print("multiply expected/decrypted:", [10.0, 40.0, 90.0, 160.0], session.decrypt(multiply_ct))
print("square expected/decrypted:", [1.0, 4.0, 9.0, 16.0], session.decrypt(square_ct))
```

## Encrypted reductions

```python
sum_ct = session.sum(left_ct)
mean_ct = session.mean(left_ct)
variance_ct = session.variance(left_ct)

print("sum expected/decrypted:", 10.0, session.decrypt(sum_ct))
print("mean expected/decrypted:", 2.5, session.decrypt(mean_ct))
print("variance expected/decrypted:", 1.25, session.decrypt(variance_ct))
```

## Close the session

```python
session.close()
print("GPU session closed")
```
