# HE SDK GPU example

This notebook uses the FIDES GPU backend installed in the notebook image. Run the cells from top to bottom.

## 1. Import the SDK and create one GPU session

```python
from he_sdk import HESession
from he_sdk_fides import FidesBackend

print("GPU plugin:", FidesBackend.__name__)
session = HESession.create(backend="fides")
print("backend:", session.capabilities.backend)
print("operations:", session.capabilities.operations)
```

## 2. Encrypt two small vectors

```python
left_values = [1.0, 2.0, 3.0, 4.0]
right_values = [10.0, 20.0, 30.0, 40.0]

left = session.encrypt(left_values)
right = session.encrypt(right_values)

print("left plaintext:", left_values)
print("right plaintext:", right_values)
print("left ciphertext metadata:", left.metadata)
```

## 3. Arithmetic operations

```python
added = session.decrypt(session.add(left, right))
subtracted = session.decrypt(session.subtract(left, right))
multiplied = session.decrypt(session.multiply(left, right))
squared = session.decrypt(session.square(left))

print("add:", added)
print("subtract:", subtracted)
print("multiply:", multiplied)
print("square:", squared)
```

## 4. Aggregate operations

```python
total = session.decrypt(session.sum(left))
average = session.decrypt(session.mean(left))
population_variance = session.decrypt(session.variance(left))

print("sum:", total)
print("mean:", average)
print("population variance:", population_variance)
```

## 5. Close the session

```python
session.close()
print("session closed")
```
